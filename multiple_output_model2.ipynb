```python
import pandas as pd
from sklearn.datasets import fetch_openml
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
```

```python
rf = fetch_openml(data_id = 41401,as_frame = True ,parser = 'auto')
x = rf.data
y  = rf.target
print(f"\nDataset loaded successfully!")
print(f"Features shape (Rows, Columns): {x.shape}")
print(f"Targets shape  (Rows, Columns): {y.shape}")
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)

#our  model
rf_model = RandomForestRegressor(n_estimators = 100,max_depth = 20,random_state =42 ,n_jobs = 1)
print("training model on large datasets..")
rf_model.fit(x_train,y_train)

y_pred = rf_model.predict(x_test)
print(f"R2 score: {r2_score(y_test,y_pred)}")

```

```python
rf_model2 = DecisionTreeRegressor( max_depth=2000,random_state=42)
print("training model on large datasets..")
rf_model2.fit(x_train,y_train)

y_pred2 = rf_model2.predict(x_test)
print(f"r2 score : {r2_score(y_test,y_pred2)}")
```
