FOR TRAINIG A MACHINE ALGORITHM FOR  CLASSIFICATION

```python
#three flower species :- iris-setosa,versicolor,virginica
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score,classification_report

```

```python
#loadin.. datasets
iris = load_iris()
x = iris.data
y = iris.target
print("dataser shape",x.shape)
print("features names",iris.feature_names)
print("flower classes ",iris.target_names)
print("first 10 features rows :\n ",x[:10])
print("first 10 target or labels :",y[:10])
```

```python
x_train , x_test,y_train,y_test = train_test_split(x,y, test_size=0.20,random_state =42,stratify=y)

#train
model = LogisticRegression(max_iter = 100)
model.fit(x_train,y_train)
print("\n model training completed!")
#testing the unseen data
y_pred = model.predict(x_test)
accuracy = accuracy_score(y_test,y_pred)
print("\n accuracy of model on unseen data is :",round(accuracy*100,4))
```

```python
#performance for classification through logistic regression
print("\n classification report")
print(classification_report(y_test,y_pred,target_names=iris.target_names))
```

```python
#same experimenting for different classification algorithms

model2 = DecisionTreeClassifier(max_depth = 3,random_state = 42)
model2.fit(x_train,y_train)

y_pred2 = model2.predict(x_test)
accuracy2 = accuracy_score(y_test,y_pred2)
print("\n accuracy of model on unseen data is :",round(accuracy2*100,4))



```

```python
#performance for classification through decision trees
print("\n classification report")
print(classification_report(y_test,y_pred2,target_names=iris.target_names))
```

```python
model3 = RandomForestClassifier(n_estimators=1000,random_state=42)
model3.fit(x_train,y_train)

y_pred3 = model3.predict(x_test)
accuracy3 = accuracy_score(y_test,y_pred3)
print("\n accuracy of model on unseen data is :",round(accuracy3*100,4))
```

```python
#performance for classification through random forest
print("\n classification report")
print(classification_report(y_test,y_pred3,target_names=iris.target_names))
```
